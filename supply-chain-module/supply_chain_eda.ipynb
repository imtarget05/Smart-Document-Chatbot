# Supply Chain Analytics EDA

## Business Metrics
- **Inventory Turnover** = COGS / Average Inventory
- **Days of Inventory On Hand (DOH)** = Average Inventory / (COGS/365)
- **Fill Rate** = Orders Fulfilled / Total Orders × 100
- **OTIF (On-Time In-Full)** = % orders delivered on time with complete quantity
- **Forecast Accuracy (MAPE)** = Mean Absolute Percentage Error
- **Supplier Lead Time** = avg delivery time from order to receipt
- **Defect Rate** = defective units / total received units

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load sample data
df = pd.read_csv('data/retail_sales.csv')

# Data overview
print(df.info())
print(df.describe())

# Time series trend
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date')
df.plot(x='date', y='sales', figsize=(12,6))
plt.title('Sales Trend')
plt.show()

# Seasonality
df['month'] = df['date'].dt.month
df.groupby('month')['sales'].mean().plot(kind='bar')
plt.title('Monthly Seasonality')
plt.show()
```
