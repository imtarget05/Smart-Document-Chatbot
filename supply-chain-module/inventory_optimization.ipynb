# Inventory Optimization

## EOQ (Economic Order Quantity)
$$Q^* = \sqrt{\frac{2DS}{H}}$$
- D = annual demand (units/year)
- S = ordering cost ($/order)
- H = holding cost ($/unit/year)

## Safety Stock
$$SS = Z \times \sigma_d \times \sqrt{L}$$
- Z = service level z-score (1.65 for 95%)
- σ_d = demand standard deviation
- L = lead time (days)

## Reorder Point
$$ROP = d \times L + SS$$
- d = average daily demand
- L = lead time (days)

```python
import numpy as np

def eoq(demand, order_cost, holding_cost):
    """Calculate Economic Order Quantity"""
    return np.sqrt(2 * demand * order_cost / holding_cost)

def safety_stock(z_score, std_demand, lead_time):
    """Calculate safety stock"""
    return z_score * std_demand * np.sqrt(lead_time)

def reorder_point(daily_demand, lead_time, safety_stock):
    """Calculate reorder point"""
    return daily_demand * lead_time + safety_stock

# Example calculation
demand = 10000  # units/year
order_cost = 50  # $/order
holding_cost = 2  # $/unit/year
z_score = 1.65  # 95% service level
std_demand = 15  # units/day std dev
lead_time = 5  # days
daily_demand = demand / 365

eoq_val = eoq(demand, order_cost, holding_cost)
ss_val = safety_stock(z_score, std_demand, lead_time)
rop_val = reorder_point(daily_demand, lead_time, ss_val)

print(f"EOQ: {eoq_val:.1f} units")
print(f"Safety Stock: {ss_val:.1f} units")
print(f"Reorder Point: {rop_val:.1f} units")

# Cost analysis
ordering_cost = (demand / eoq_val) * order_cost
holding_cost_total = (eoq_val / 2) * holding_cost
total_cost = ordering_cost + holding_cost_total

print(f"\nAnnual Ordering Cost: ${ordering_cost:.2f}")
print(f"Annual Holding Cost: ${holding_cost_total:.2f}")
print(f"Total Cost: ${total_cost:.2f}")
```
