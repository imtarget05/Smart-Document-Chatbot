# Anomaly Detection for Supply Chain

## Use Cases
- Demand spike detection (unusual sales surge)
- Inventory level anomaly (unexpected stock changes)
- Supplier delay anomaly (late deliveries beyond threshold)
- Procurement fraud detection (unusual order patterns)

## Methods
1. Isolation Forest (unsupervised)
2. Statistical control charts (3-sigma rule)
3. DBSCAN clustering (density-based)
4. One-Class SVM (optional)

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import IsolationForest

# Load demand data
df = pd.read_csv('data/demand_history.csv')
df['date'] = pd.to_datetime(df['date'])

# Select features for anomaly detection
features = ['demand', 'inventory_level', 'lead_time_days', 'defect_rate']
X = df[features].dropna()

# Isolation Forest model
model = IsolationForest(contamination=0.05, random_state=42)
model.fit(X)

# Predict anomalies
df['anomaly'] = model.predict(X)  # -1 = anomaly, 1 = normal
df['anomaly_score'] = model.decision_function(X)

# Count anomalies
n_anomalies = (df['anomaly'] == -1).sum()
total = len(df)
print(f"Anomalies detected: {n_anomalies} out of {total} ({n_anomalies/total*100:.2f}%)")

# Show top anomalous records
anomalies = df[df['anomaly'] == -1].sort_values('anomaly_score')
print("\nTop 10 anomalies:")
print(anomalies[['date', 'demand', 'inventory_level', 'lead_time_days', 'defect_rate', 'anomaly_score']].head(10))
```
