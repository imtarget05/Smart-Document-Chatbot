# Demand Forecasting

Time series forecasting for retail sales demand.

## Evaluation Metrics
- **MAPE** = Mean Absolute Percentage Error
- **RMSE** = Root Mean Squared Error
- **sMAPE** = Symmetric MAPE

## Models Compared
1. Naive (last value)
2. Moving Average
3. Prophet
4. ARIMA (optional)

```python
import pandas as pd
import numpy as np
from prophet import Prophet
from sklearn.metrics import mean_absolute_error, mean_squared_error

df = pd.read_csv('data/retail_sales.csv')
df['ds'] = pd.to_datetime(df['date'])
df['y'] = df['sales']

# Train/test split
train = df[df['ds'] < '2023-01-01']
test = df[df['ds'] >= '2023-01-01']

# Prophet model
model = Prophet()
model.fit(train[['ds', 'y']])

# Forecast
future = model.make_future_dataframe(periods=len(test))
forecast = model.predict(future)

# Evaluate
mape = mean_absolute_error(test['y'], forecast['yhat'].iloc[-len(test):]) / test['y'].mean() * 100
rmse = np.sqrt(mean_squared_error(test['y'], forecast['yhat'].iloc[-len(test):]))

print(f'MAPE: {mape:.2f}%')
print(f'RMSE: {rmse:.2f}')

# Visualization
model.plot(forecast)
plt.show()
```
